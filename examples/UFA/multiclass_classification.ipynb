```python
from sklearn.datasets import load_iris
from dynamodelx import UFA
from dynamodelx.plots import draw_plots
```

```python
data = load_iris()
X, y = data.data, data.target
classes = data.target_names
```

```python
ufa = UFA(
    task='classification', 
    model_size=None, # None because I want to use my own architecture. 
    input_dim=X.shape[1],
    output_dim=len(classes),
    loss='cross_entropy_loss',
    device='cuda',
    custom_architecture=[6],
    weights_init='xavier',
    hidden_activation='tanh',
    optimizer='adam',
    return_metrics=True,
    auto_build=True,
    multiclass=True, # Since it's multi-class classification
    uncertainty=False
)

performance = ufa.train(
                    X=X,
                    y=y,
                    epochs=50,
                    learning_rate=0.01,
                    momentum=None,
                    val_size=0.2,
                    test_size=0.1,
                    batch_size=32
                )
```

```python
performance.test
```

```python
draw_plots(performance)
```

```python
predictions = ufa.predict(X[:5])
print(predictions, all(predictions == y[:5]))
```

```python
ufa.save(parameters_path='iris_classification_params.pt', arguments_path='iris_classification_args.pt')
```

```python
ufa = UFA.load(parameters_path='iris_classification_params.pt', arguments_path='iris_classification_args.pt')
predictions = ufa.predict(X[:5])
print(predictions, all(predictions == y[:5]))
```
